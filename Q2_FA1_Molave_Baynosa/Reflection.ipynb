1. What functions did you create in your calculator program?
I wrote four functions, one for each basic operation: add_numbers, subtract_numbers, multiply_numbers, and divide_numbers.

2. What parameters did your functions use?
Every function took two parameters, num1 and num2, which hold the two numbers the calculation is performed on.

3. What arguments were passed when the functions were called?
The two numbers the user typed in were passed as arguments. Which function got called depended on the operation the user picked from the menu.

4. How did your program use the returned value?
The function returned its result, and the program displayed that value as the final answer.

5. Why is it better to divide the calculator program into functions instead of writing everything in one long block of code?
Splitting the program into functions keeps it neat, readable, and easier to troubleshoot. Each function handles a single job, so I can fix or change one operation without touching the rest of the program.
