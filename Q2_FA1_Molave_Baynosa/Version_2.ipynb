|Division|divide_numbers|num1, num2|divide the two numbers|Quotient|

```python
# Modular Calculator Mini-Program
import operator

# Menu choice -> (name, function)
OPERATIONS = {
    "1": ("Addition", operator.add),
    "2": ("Subtraction", operator.sub),
    "3": ("Multiplication", operator.mul),
    "4": ("Division", operator.truediv),
}


def get_number(prompt):
    # Ask until the user types a valid number
    while True:
        try:
            return float(input(prompt))
        except ValueError:
            print("Please enter a valid number.")


def show_menu():
    # Print the menu from OPERATIONS
    print("\nChoose operation:")
    for key, (label, _) in OPERATIONS.items():
        print(f"{key} - {label}")
    print("q - Quit")


def calculate(a, b, op):
    # Do the chosen math
    return OPERATIONS[op][1](a, b)


def main():
    while True:  # Keep going until quit
        show_menu()
        choice = input("Enter choice: ").strip().lower()

        if choice == "q":
            print("Goodbye!")
            break
        if choice not in OPERATIONS:
            print("Invalid choice.")
            continue

        x = get_number("Enter first number: ")
        y = get_number("Enter second number: ")

        try:
            result = calculate(x, y, choice)
            print(f"Result: {result:g}")  # Show 5 instead of 5.0
        except ZeroDivisionError:
            print("Cannot divide by zero.")


if __name__ == "__main__":  # Run only when started directly
    main()
```
